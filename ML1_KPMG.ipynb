```python

# Step 1: Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import joblib
```

```python

# Step 2: Define features (X) and target (y)

target = "Annual_Revenue"

drop_columns = [
    "Customer_ID",
    "Customer_Name",
    "Annual_Revenue",
    "Customer_Lifetime_Value",
    "Customer_Value_Segment"
]

X = df.drop(columns=drop_columns)
y = df[target]

print("Feature dataset shape:", X.shape)
print("Target dataset shape:", y.shape)

print("\nTarget variable summary:")
print(y.describe())
```

```python

# Step 3: Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python

# Step 4: Identify column types

numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical features:", numeric_features)
print("\nCategorical features:", categorical_features)
```

```python

# Step 4.1: Preprocess numerical features

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Step 4.2: Preprocess categorical features

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Step 4.3: Combine both transformers

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully.")
```

```python

# Step 5: Define regression models

models = {
    "Linear Regression": LinearRegression(),

    "Ridge Regression": Ridge(alpha=1.0),

    "Lasso Regression": Lasso(alpha=0.1, max_iter=10000),

    "Decision Tree": DecisionTreeRegressor(
        random_state=42,
        max_depth=15
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        random_state=42,
        n_jobs=-1,
        max_depth=20
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=150,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    )
}

print("Total models to train:", len(models))
```

```python

# Step 6: Train and evaluate all regression models

results = []
trained_models = {}

for name, model in models.items():

    # Create a complete pipeline for each model
    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("regressor", model)
    ])

    # Train the model
    pipeline.fit(X_train, y_train)

    # Predict annual revenue on unseen test data
    y_pred = pipeline.predict(X_test)

    # Calculate evaluation metrics
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)

    # Store results
    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2 Score": r2
    })

    trained_models[name] = pipeline

    print(f"{name} trained successfully.")

# Create the comparison table
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="RMSE",
    ascending=True
).reset_index(drop=True)

print("\nRegression Model Comparison:")
print(results_df.round(4).to_string(index=False))
```

```python

# Step 7: Compare models visually

plt.figure(figsize=(10, 5))

plt.bar(
    results_df["Model"],
    results_df["RMSE"]
)

plt.title("Regression Models - RMSE Comparison")
plt.xlabel("Regression Model")
plt.ylabel("RMSE")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

```python

# Compare R2 scores

plt.figure(figsize=(10, 5))

plt.bar(
    results_df["Model"],
    results_df["R2 Score"]
)

plt.title("Regression Models - R2 Score Comparison")
plt.xlabel("Regression Model")
plt.ylabel("R2 Score")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

```python

# Step 8: Select the best model based on test RMSE

best_model_name = results_df.loc[0, "Model"]
best_model = trained_models[best_model_name]

print("Best model:", best_model_name)
print("MAE:", round(results_df.loc[0, "MAE"], 2))
print("RMSE:", round(results_df.loc[0, "RMSE"], 2))
print("R2 Score:", round(results_df.loc[0, "R2 Score"], 4))

# Save the complete pipeline
joblib.dump(
    best_model,
    "urbankart_best_revenue_model.pkl"
)

print("\nBest model saved successfully.")
```

```python

# Step 9: Five-fold cross-validation on training data

top_models = results_df["Model"].head(3).tolist()

cv_results = []

for name in top_models:

    pipeline = trained_models[name]

    scores = cross_val_score(
        pipeline,
        X_train,
        y_train,
        cv=5,
        scoring="neg_root_mean_squared_error",
        n_jobs=-1
    )

    fold_rmse = -scores

    cv_results.append({
        "Model": name,
        "Mean CV RMSE": fold_rmse.mean(),
        "Std CV RMSE": fold_rmse.std()
    })

cv_results_df = pd.DataFrame(cv_results)

cv_results_df = cv_results_df.sort_values(
    by="Mean CV RMSE"
).reset_index(drop=True)

print(cv_results_df.round(2).to_string(index=False))
```
